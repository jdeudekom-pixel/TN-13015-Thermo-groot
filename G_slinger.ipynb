## Processing
**Description of processing of raw data into scientific evidence:**


````python
#Data processing and analysis:

def bereken_g(T, L):
    g = (4 * np.pi**2 * L) / (T**2)
    return g


def bereken_dT(T):
    return np.std(T) / np.sqrt(len(T))


def bereken_dg(T, L, dL):
    dT = bereken_dT(T)
    g = bereken_g(np.mean(T), L)
    return g * np.sqrt((dL / L) ** 2 + (2 * dT / np.mean(T)) ** 2)


# meting = metingen_L3[0]
# meting = T_tot_L2
# l = L2

# print(f"g = {bereken_g(np.mean(meting), l):.4f} ± {bereken_dg(meting, l, dL):.4f} m/s²")
# print(f"error = {bereken_dg(meting, l, dL) / bereken_g(np.mean(meting), l) * 100:.4f}%")

lengtes = np.array([L2, L3, L4, L5, L6, L7])
Ts = np.array(
    [
        np.mean(T_tot_L2),
        np.mean(T_tot_L3),
        np.mean(T_tot_L4),
        np.mean(T1L5),
        np.mean(T1L6),
        np.mean(T1L7),
    ]
)
Ts2 = Ts**2


def lineair(L, a, b):
    return a * (L + b)


popt, pcov = curve_fit(lineair, lengtes, Ts2, p0=[4 * np.pi**2 / 9.812, 0])
g = 4 * np.pi**2 / popt[0]
print(g, np.sqrt(pcov[0, 0]))
````
